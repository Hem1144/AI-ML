## Object detection in an image

```python
# Install the Ultralytics YOLOv8 library using pip
! pip install ultralytics
# ! pip install opencv-python
```

```python
# Import the YOLO class from the Ultralytics library for object detection
from ultralytics import YOLO

# Import OpenCV for image processing
import cv2

# Import Matplotlib for displaying images and visualizations
import matplotlib.pyplot as plt
```

```python
# Load YOLOv8 model (you can use 'yolov8n.pt', 'yolov8s.pt', 'yolov8m.pt', 'yolov8l.pt', or 'yolov8x.pt')
model = YOLO("yolov8n.pt")  # Python API, not CLI

# Load and convert the image
# Please upload your image
image_path = "/content/drive/MyDrive/ai_labs/horse.jpg"
image_bgr = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# Run object detection
results = model.predict(source=image_rgb, save=False)
```

```python
# Visualize the result
annotated_image = results[0].plot()

# Display the image with bounding boxes
plt.imshow(annotated_image)
plt.axis('off')
plt.title("YOLOv8 Detection Result")
plt.show()
```

## Object Tracking in a Video

```python
# Import the files module from Google Colab to handle file uploads and downloads
from google.colab import files
```

```python
# Load the YOLOv8 model
model = YOLO("yolov8n.pt")

# Load the video file
video_path = "/content/drive/MyDrive/ai_labs/people-walking.mp4"
cap = cv2.VideoCapture(video_path)

# Check if video opened successfully
if not cap.isOpened():
    raise IOError(f"Cannot open video: {video_path}")

# Get video properties
frame_width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps          = cap.get(cv2.CAP_PROP_FPS)

# Save output video
out = cv2.VideoWriter("/content/drive/MyDrive/ai_labs/output_yolo.mp4",
                      cv2.VideoWriter_fourcc(*'mp4v'),
                      fps,
                      (frame_width, frame_height))

# Process video
while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(source=frame, save=False, conf=0.3)
    annotated_frame = results[0].plot()
    out.write(annotated_frame)

# Clean up
cap.release()
out.release()


files.download("output_yolo.mp4")
```

## Additional practice: Object tracking for desktop environments such as VS Code

```python
# pip install ultralytics
# pip install opencv-python

from ultralytics import YOLO
import cv2

# Load model
model = YOLO("yolov8n.pt")

# Open video
video_path = "people-walking.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise IOError(f"Cannot open video: {video_path}")

# Get video properties
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Save output (optional)
save_output = True
if save_output:
    out = cv2.VideoWriter("output_tracking.mp4",
                          cv2.VideoWriter_fourcc(*'mp4v'),
                          fps,
                          (frame_width, frame_height))

# Tracking and display loop
while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model.track(frame, persist=True, conf=0.3, tracker="bytetrack.yaml")
    annotated_frame = results[0].plot()

    # Show frame live
    cv2.imshow("YOLOv8 Tracking", annotated_frame)

    if save_output:
        out.write(annotated_frame)

    # Press 'q' to stop
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
if save_output:
    out.release()
cv2.destroyAllWindows()
```
